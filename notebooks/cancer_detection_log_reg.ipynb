# Breast Cancer Classification using Logistic Regression
Exploratory training, evaluation, and model export workflow.

```python
# Importing the libraries
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
```

```python
# Importing the dataset
dataset = pd.read_csv('../data/cancer_data.csv')
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

```python
# Splitting into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# Training the Logistic Regression model
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state=0, max_iter=1000)
classifier.fit(X_train, y_train)
```

```python
# Predicting the test set results & evaluating performance
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.model_selection import cross_val_score

y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print('Confusion Matrix:\n', cm)
print(f'Test Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%')

accuracies = cross_val_score(estimator=classifier, X=X_train, y=y_train, cv=10)
print('10-Fold CV Accuracy: {:.2f} %'.format(accuracies.mean()*100))
print('Standard Deviation: {:.2f} %'.format(accuracies.std()*100))
```

```python
# Save the trained model
os.makedirs('../models', exist_ok=True)
joblib.dump(classifier, '../models/logistic_model.pkl')
print('Model saved to ../models/logistic_model.pkl')
```

```python
# Load the saved model and generate predictions on new dataset
loaded_classifier = joblib.load('../models/logistic_model.pkl')
new_dataset = pd.read_csv('../data/new_datasets.csv')
X_new = new_dataset.values

new_preds = loaded_classifier.predict(X_new)
print('Predicted Output:\n', new_preds)

df_with_predictions = new_dataset.copy()
df_with_predictions['Predicted_Target'] = new_preds
df_with_predictions.to_csv('../data/new_dataset_with_predictions.csv', index=False)
print("Predictions saved to '../data/new_dataset_with_predictions.csv'")
```
